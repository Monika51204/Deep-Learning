## 5. Training 5 models

```python
from ml_models import build_model, train_one_model, predict_price


# Xác định các ID mô hình
model_ids = ["MODEL_01","MODEL_02","MODEL_03","MODEL_04","MODEL_05"]

# Hàm lấy parameters của model
def get_model_params(model_id):
    model = build_model(model_id)

    params = model.get_params()

    # Giữ parameter có giá trị khác None
    params = {
        key: value
        for key, value in params.items()
        if value is not None
    }

    return params

# Training và evaluation
results = []

for model_id in model_ids:
    # Lấy parameters
    params = get_model_params(model_id)

    # Train model
    model, training_time_seconds = train_one_model(
        model_id,
        X_train,
        y_train
    )

    # Predict validation set
    val_pred = predict_price(
        model,
        X_val
    )

    # Tính metrics
    metrics = regression_metrics(
        y_val,
        val_pred
    )

    # Print kết quả
    print("=" * 80)
    print("MODEL:", model_id)
    print("Parameters:", params)
    print("Metrics:", metrics)
    print("Training time:", training_time_seconds)
    print("=" * 80)

    # Save kết quả
    results.append({
        "model": model_id,
        "parameters": params,
        "metrics": metrics,
        "training_time_seconds": training_time_seconds,
        "kaggle_score": None,
        "notes": ""
    })
```

## 6. Thử các mô hình bằng tham số khác


```python
import random
#Load lại module ml_models để áp dụng các thay đổi mới nhất
importlib.reload(ml_models)

# Tạo các hyperparameters ngẫu nhiên cho các mô hình khác nhau
def random_hyperparameters(model_id):

    if model_id == "MODEL_02":

        return {
            "max_depth": random.choice([None, 5, 10, 15, 20, 25, 30]),
            "min_samples_split": random.choice([2, 5, 10, 15, 20]),
            "min_samples_leaf": random.choice([1, 2, 4, 8, 10])
        }


    elif model_id == "MODEL_03":

        return {
            "n_estimators": random.choice([100, 200, 300, 400, 500]),
            "max_depth": random.choice([None, 5, 10, 15, 20]),
            "min_samples_split": random.choice([2, 5, 10, 15]),
            "min_samples_leaf": random.choice([1, 2, 4, 8])
        }


    elif model_id == "MODEL_04":

        return {
            "n_estimators": random.choice([100, 200, 300, 400, 500]),
            "learning_rate": random.choice([
                0.01, 0.03, 0.05, 0.1
            ]),
            "max_depth": random.choice([2, 3, 4, 5]),
            "min_samples_split": random.choice([2, 5, 10, 15]),
            "min_samples_leaf": random.choice([1, 2, 4, 8])
        }


    elif model_id == "MODEL_05":

        return {
            "n_estimators": random.choice([100, 200, 300, 500, 700]),
            "learning_rate": random.choice([
                0.01, 0.03, 0.05, 0.1
            ]),
            "max_depth": random.choice([3, 4, 5, 6, 8]),
            "min_child_weight": random.choice([1, 3, 5]),
            "subsample": random.choice([
                0.7, 0.8, 0.9, 1.0
            ]),
            "colsample_bytree": random.choice([
                0.7, 0.8, 0.9, 1.0
            ])
        }

    else:

        raise ValueError(
            f"Random hyperparameters are not defined "
            f"for {model_id}"
        )
```

```python
model_ids = ["MODEL_01", "MODEL_02", "MODEL_03", "MODEL_04", "MODEL_05"]

results = []

for model_id in model_ids:

    # Lấy object model ra mà không sửa hàm gốc
    model, training_time_seconds = train_one_model(
        model_id,
        X_train,
        y_train
    )

    # Lấy params ngẫu nhiên
    try:
        params = random_hyperparameters(model_id)
    except ValueError:
        params = {}

    # Gán đè params ngẫu nhiên vào model và fit lại để tính lại thời gian chuẩn
    if params:
        model.set_params(**params)
        
        start_time = time.time()
        model.fit(X_train, y_train)
        training_time_seconds = time.time() - start_time

    # In hyperparameters
    print("=" * 80)
    print(f"MODEL: {model_id}")
    print("Parameters:", {
        key: value
        for key, value in model.get_params().items()
        if key in [
            "n_estimators",
            "max_depth",
            "learning_rate",
            "min_samples_split",
            "min_samples_leaf",
            "min_child_weight",
            "subsample",
            "colsample_bytree",
            "random_state",
            "n_jobs"
        ]
    })

    # Predict validation
    val_pred = predict_price(
        model,
        X_val
    )

    # Evaluation
    metrics = regression_metrics(
        y_val,
        val_pred
    )

    # Print results
    print()
    print("Metrics:")
    print(f"  RMSLE: {metrics['RMSLE']:.5f}")
    print(f"  RMSE:  {metrics['RMSE']:.2f}")
    print(f"  MAE:   {metrics['MAE']:.2f}")
    print(f"  R2:    {metrics['R2']:.5f}")
    print(f"  Training Time: {training_time_seconds:.4f} seconds")
    print("=" * 80)
    print()

    # Lưu kết quả 
    results.append({
        "model": model_id,
        "metrics": metrics,
        "training_time_seconds": training_time_seconds,
        "parameters": params,
        "kaggle_score": None,
        "notes": ""
    })
```
