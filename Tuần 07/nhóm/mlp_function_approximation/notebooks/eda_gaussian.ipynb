```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Notebook nằm trong thư mục notebooks/
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

data_path = ROOT / "data" / "processed" / "train_pool.csv"

print("Dataset path:", data_path)

if data_path.exists():
    df = pd.read_csv(data_path)

    display(df.head())
    display(df.describe())

    ax = df.plot.scatter(
        x="x1",
        y="x2",
        c="y",
        colormap="viridis"
    )
    ax.set_title("Train pool: x1 vs x2 (color=y)")
    plt.show()
else:
    print("Dataset not found yet.")
```
